```python
for file in sales_files:
    df = pd.read_excel(file)

    print(
        file.name,
        df.shape,
        len(df.columns)
    )
```

```python
base_columns = set(
    pd.read_excel(sales_files[0], nrows=0).columns
)

for file in sales_files[1:]:
    current_columns = set(
        pd.read_excel(file, nrows=0).columns
    )

    if current_columns != base_columns:
        print("Different schema:", file.name)
```

```python
sales_list = []

for file in sales_files:

    print("Reading:", file.name)

    df = pd.read_excel(file)

    df["source_file"] = file.name

    sales_list.append(df)

sales = pd.concat(
    sales_list,
    ignore_index=True
)
```

```python
quality_report = pd.DataFrame({
    "dtype": sales.dtypes.astype(str),
    "rows": len(sales),
    "missing": sales.isnull().sum(),
    "missing_pct": (
        sales.isnull().mean() * 100
    ).round(2),
    "unique": sales.nunique()
})

quality_report
```
